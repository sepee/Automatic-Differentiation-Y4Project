# Topological sorting

Here is an example to test how to topologically sort nodes

```python
topo_order = []

class node:
    def __init__(self, _name, _children):
        self.name = _name
        self.children = _children
        self.visited = False
        self.nextChild = None

    def sort(self, order):
        if self.visited:
            return

        for child in self.children:
            child.sort(order)

        if len(order) > 0:
            self.nextChild = order[0]
        order.insert(0, self)
        self.visited = True

    def __call__(self):
        if not self.nextChild == None:
            return f"{self.name}, {self.nextChild()}"
        return self.name

    def eval_depth_first(self):
        print(self.name)
        for child in self.children:
            child.eval_depth_first()


a = node("a", [])
b = node("b", [a])
c = node("c", [a])
d = node("d", [b, c])
e = node("e", [a, d])
f = node("f", [c, e])

f.sort(topo_order)

print("order:")
print(f())

print("depth-first:")
f.eval_depth_first()
```
